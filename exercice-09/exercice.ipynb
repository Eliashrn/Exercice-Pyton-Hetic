```python
import pandas as pd

df = pd.read_csv("../data/titanic_propre.csv")
df.shape 
```

```python
df["tranche"] = pd.cut(
    df["age"],
    bins=[0, 12, 17, 34, 59, 100],
    labels=["0-12", "13-17", "18-34", "35-59", "60+"],
)
(df.groupby("tranche", observed=True)["survived"].mean() * 100).round(1)
```

```python
df.groupby("embark_town")["fare"].mean().round(2).sort_values(ascending=False)
```

```python
df["seul"] = (df["sibsp"] + df["parch"]) == 0
(df.groupby("seul")["survived"].mean() * 100).round(1)
(df.groupby(["sex", "seul"])["survived"].mean() * 100).round(1).unstack()
```
