```python
import seaborn as sns
import pandas as pd
titanic = sns.load_dataset("titanic")
titanic.head(10)
```

```python
mineurs = titanic[titanic["age"] < 18]
print(mineurs)
```

Le nombre de passager mineur est de 113

```python
ticket = titanic[["fare", "embark_town", "class"]].sort_values("fare", ascending=False).head(10)
print(ticket)
```

```python
women = titanic[titanic["sex"] == "female"].value_counts()
sex = titanic["sex"].value_counts()
print(women * sex / 100)
```

```python
premiere = titanic[titanic["pclass"] == 1]
pourcentage = (premiere["sex"] == "female").mean() * 100
print(pourcentage)
```

```python
femmes_1 = titanic[(titanic["pclass"] == 1) & (titanic["sex"] == "female")]
taux_f1 = femmes_1["survived"].mean() * 100
print(f"Femmes 1re classe : {taux_f1:.1f} %")

```

```python
femmes_1 = titanic.query("pclass == 1 and sex == 'female'")
hommes_3 = titanic.query("pclass == 3 and sex == 'male'")

print(femmes_1["survived"].mean() * 100)
print(hommes_3["survived"].mean() * 100)
```
